```python
import pandas as pd

data = {
    "customer_id": [1, 2, 3, 4, 5],
    "signup_date": ["2024-01-10", "2024-02-05", "2024-03-12", "2024-04-01", "2024-05-20"],
    "last_purchase": ["2024-02-15", "2024-03-01", "2024-03-28", "2024-05-10", "2024-06-25"]
}

df = pd.DataFrame(data)
```

```python
print(df.dtypes)
```

```python
df["signup_date"] = pd.to_datetime(df["signup_date"])
df["last_purchase"] = pd.to_datetime(df["last_purchase"])

print(df.dtypes)
```

```python
df["signup_year"] = df["signup_date"].dt.year
df["signup_month"] = df["signup_date"].dt.month
df["signup_day"] = df["signup_date"].dt.day
df["signup_quarter"] = df["signup_date"].dt.quarter
df["signup_month_name"] = df["signup_date"].dt.month_name()
df["signup_day_name"] = df["signup_date"].dt.day_name()
```

```python
print(df)
```

```python
df["signup_dow"] = df["signup_date"].dt.dayofweek
df["is_weekend"] = df["signup_dow"].apply(lambda x: 1 if x >= 5 else 0)
```

```python
df["days_between_signup_and_purchase"] = (
    df["last_purchase"] - df["signup_date"]
).dt.days
```

```python
after_march = df[df["signup_date"] > "2024-03-01"]
march = df[df["signup_date"].dt.month == 3]
year_2024 = df[df["signup_date"].dt.year == 2024]
```

```python
some_date = pd.Timestamp("2024-08-18")

print(f"Original date: {some_date}")
print(f"Beginning of month: {some_date.to_period('M').start_time}")
print(f"End of month: {some_date.to_period('M').end_time}")
```

```python
df["month_start"] = df["signup_date"].dt.to_period("M").dt.start_time
df["month_end"] = df["signup_date"].dt.to_period("M").dt.end_time.dt.normalize()
```
